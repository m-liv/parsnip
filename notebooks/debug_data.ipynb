```python
def load_boolq(dialect, n, seed):
    split = ds[dialect].shuffle(seed=seed).select(range(n))
    
    examples = []
    
    for e in split:
        examples.append({
            "sae_passage": e["SAE Passage"],
            "sae_question": e["SAE Question"],
            "dialect_passage": e["Dialect (SAE Passage)"],
            "label": e["Actual Label"],
            "bleu": e["BLEU Score SAE Passage"]
        })
        
    return examples

def pretty_print_boolq(example, dialect_name):
    print("=" * 100)
    print(f"DIALECT SPLIT: {dialect_name}")
    print("-" * 100)
    
    print("SAE PASSAGE:")
    print(example["sae_passage"])
    print()
    
    print("DIALECT PASSAGE:")
    print(example["dialect_passage"])
    print()
    
    print("QUESTION:")
    print(example["sae_question"])
    print()
    
    print("LABEL:")
    print(example["label"])
    print("=" * 100)
    print("\n\n")
```

```python
examples = load_boolq("AAVE", N, SEED)
print(len(examples))
for e in examples:
    pretty_print_boolq(e, "AAVE")
```

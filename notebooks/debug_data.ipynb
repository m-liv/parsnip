```python
def load_boolq(dialect, n, seed):
    ds = load_dataset("abhaygupta1266/boolq")
    split = ds[dialect].shuffle(seed=seed).select(range(n))
    
    examples = []
    
    for e in split:
        examples.append({
            "sae_passage": e["SAE Passage"],
            "sae_question": e["SAE Question"],
            "dialect_passage": e["Dialect (SAE Passage)"],
            "label": e["Actual Label"],
            "bleu": e["BLEU Score SAE Passage"]
        })
        
    return examples

def pretty_print_boolq(example, dialect_name):
    print("=" * 100)
    print(f"BOOLQ - DIALECT SPLIT: {dialect_name}")
    print("-" * 100)
    
    print("SAE PASSAGE:")
    print(example["sae_passage"])
    print()
    
    print("DIALECT PASSAGE:")
    print(example["dialect_passage"])
    print()
    
    print("QUESTION:")
    print(example["sae_question"])
    print()
    
    print("LABEL:")
    print(example["label"])
    print()
    
    print("BLEU SCORE:")
    print(example["bleu"])
    
    print("=" * 100)
    print("\n\n")
```

```python
test_N = 5
```

```python
boolq_examples = load_boolq("AAVE", test_N, SEED)
print(len(boolq_examples))
for e in boolq_examples:
    pretty_print_boolq(e, "AAVE")
```

```python
def load_mbpp(dialect, n, seed):
    ds = load_dataset("abhaygupta1266/mbpp")
    split = ds[dialect].shuffle(seed=seed).select(range(n))
    
    examples = []
    for e in split:
        examples.append({
            "sae_prompt": e["Original"],
            "dialect_prompt": e["Dialect (Original)"],
            "reference_code": e["Code"],
            "test_cases": e["Test_Cases"],
            "bleu": e["BLEU Score Original"],
        })
    return examples

def pretty_print_mbpp(example, dialect_name):  
    print("=" * 100)
    print(f"MBPP - DIALECT SPLIT: {dialect_name}")
    print("-" * 100)
    
    print("SAE PROMPT:")
    print(example["sae_prompt"])
    print()
    
    print("DIALECT PROMPT:")
    print(example["dialect_prompt"])
    print()
    
    print("REFERENCE CODE:")
    print(example["reference_code"])
    print()
    
    print("TEST CASES:")
    print(example["test_cases"])
    print()
    
    print("BLEU SCORE:")
    print(example["bleu"])
    
    print("=" * 100)
    print("\n\n")
```

```python
mbpp_examples = load_mbpp("AAVE", test_N, SEED)
print(len(mbpp_examples))
for e in mbpp_examples:
    pretty_print_mbpp(e, "AAVE")
```
